```python
import cv2
import numpy as np

def dibujar_cuadricula(imagen, nx, ny, w, color=(0, 0, 255), output_path=None):
    """
    Dibuja una cuadrícula sobre una imagen.

    Parámetros:
        imagen: ruta de la imagen (str) o array numpy (imagen ya cargada).
        nx: número de celdas horizontales (columnas).
        ny: número de celdas verticales (filas).
        w: grosor de las líneas en píxeles.
        color: color de las líneas en formato BGR (tupla), por defecto rojo (0,0,255).
        output_path: ruta donde guardar la imagen resultante. Si es None, se muestra.
    """
    # Cargar la imagen si se pasó como ruta
    if isinstance(imagen, str):
        img = cv2.imread(imagen)
        if img is None:
            raise ValueError("No se pudo cargar la imagen. Verifica la ruta.")
    else:
        img = imagen.copy()  # trabajar sobre una copia

    alto, ancho = img.shape[:2]

    # Calcular el tamaño de cada celda
    ancho_celda = ancho / nx
    alto_celda = alto / ny

    # Dibujar líneas verticales (excepto el borde derecho)
    for i in range(1, nx):
        x = int(i * ancho_celda)
        cv2.line(img, (x, 0), (x, alto), color, w)

    # Dibujar líneas horizontales (excepto el borde inferior)
    for j in range(1, ny):
        y = int(j * alto_celda)
        cv2.line(img, (0, y), (ancho, y), color, w)

    # Guardar o mostrar
    if output_path:
        cv2.imwrite(output_path, img)
        print(f"Imagen guardada en: {output_path}")
    else:
        cv2.imshow('Cuadrícula', img)
        cv2.waitKey(0)
        cv2.destroyAllWindows()

    return img



    
```

```python
ruta_imagen = "sprites/jorge.png"   # Cambia por tu imagen
nx = 8          # número de columnas
ny = 3         # número de filas
grosor = 4      # grosor de línea en píxeles
color_lineas = (255, 0, 0)   # azul en BGR
ruta_salida = "sprites/imagen_con_cuadricula.jpg"

# Dibujar y guardar
resultado = dibujar_cuadricula(ruta_imagen, nx, ny, grosor, color_lineas, ruta_salida)
```
